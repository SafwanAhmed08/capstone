```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import os
import joblib
```

```python
df = pd.read_csv("../dataset/processed_dataset.csv",low_memory=False)
df = df.dropna()
```

```python
attack_labels = df['Attack_label']
attack_types = df['Attack_type']
```

```python
features = df.drop(columns=["Attack_label", "Attack_class", "Attack_type"])
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(features)
joblib.dump(scaler,"standard_scaler.pkl")
```

```python
X_benign = X_scaled[attack_labels==0]
X_attack = X_scaled[attack_labels==1]
```

```python
y_attack_types = attack_types[attack_labels==1]
```

```python
X_train_benign, X_temp = train_test_split(X_benign,test_size=30,random_state=42)
X_val_benign, X_test_benign = train_test_split(X_temp,test_size=0.67,random_state=42)
```

```python
X_val_attack, X_test_attack = train_test_split(X_attack, test_size=0.8, random_state=42)
y_val_attack, y_test_attack = train_test_split(y_attack_types, test_size=0.8, random_state=42)
```

```python
print(f"Benign splits - Train: {len(X_train_benign)}, Val: {len(X_val_benign)}, Test: {len(X_test_benign)}")
print(f"Attack splits - Val: {len(X_val_attack)}, Test: {len(X_test_attack)}")
```

```python
save_path = "./models"
```

```python
os.makedirs("./models/", exist_ok=True)
```

```python
np.save(f"{save_path}/X_train_benign.npy", X_train_benign)
np.save(f"{save_path}/X_val_benign.npy", X_val_benign)
np.save(f"{save_path}/X_test_benign.npy", X_test_benign)
np.save(f"{save_path}/X_val_attack.npy", X_val_attack)
np.save(f"{save_path}/X_test_attack.npy", X_test_attack)
```

```python
joblib.dump(scaler, f"{save_path}/standard_scaler.pkl")
```

```python
# Save attack type labels
np.save(f"{save_path}/y_val_attack.npy", y_val_attack)
np.save(f"{save_path}/y_test_attack.npy", y_test_attack)
```
